When Azure AI Services sends back a response, it doesn't come as a single value. It comes as **structured data** — a collection of results, scores, and metadata organized in a specific way.

To work with this data, you need two key Python structures:

```
┌─────────────────────────────────────────────────┐
│              Azure AI Response                   │
│                                                  │

│   │ ─────────── │    │ ─────────────────── │    │
```

## 📌 2.1 — Lists: Ordered Collections

A **list** holds multiple items in a specific order. In AI-102, you'll get lists of detected objects, languages, key phrases, entities, and more.

````python
# Creating a list
detected_languages = ["English", "French", "Spanish"]
print(detected_languages)
````

````python
# Accessing items by position (index) — starts at 0!
detected_languages = ["English", "French", "Spanish"]

print(detected_languages[0])   # First item
print(detected_languages[1])   # Second item
print(detected_languages[-1])  # Last item
````

````python
# How many items?
detected_languages = ["English", "French", "Spanish"]
print(f"Detected {len(detected_languages)} languages")
````

````python
# Adding and removing items
tags = ["outdoor", "nature"]
print(f"Before: {tags}")

tags.append("landscape")       # Add to end
print(f"After append: {tags}")

tags.remove("nature")          # Remove specific item
print(f"After remove: {tags}")
````

````python
# Checking if something is in a list
supported_languages = ["en", "fr", "de", "es", "it"]
user_language = "fr"

if user_language in supported_languages:
    print(f"✅ {user_language} is supported!")
else:
    print(f"❌ {user_language} is NOT supported")
````

## 📌 2.2 — Looping Through Lists

In AI-102 labs, you'll constantly loop through results — printing each detected entity, processing each image tag, etc.

````python
# The for loop — iterate through each item
key_phrases = ["cloud computing", "artificial intelligence", "Azure services"]

print("Key Phrases Found:")
for phrase in key_phrases:
    print(f"  • {phrase}")
````

````python
# Loop with index using enumerate()
# Useful when you need both the position AND the value
results = ["positive", "neutral", "negative", "positive"]

for i, sentiment in enumerate(results):
    print(f"  Review {i + 1}: {sentiment}")
````

````python
# Real AI-102 pattern: Loop through detected objects with confidence
detections = [
    ("person", 0.98),
    ("car", 0.87),
    ("tree", 0.72)
]

for name, confidence in detections:
    print(f"  {name}: {confidence:.0%}")
````

## 📌 2.3 — Dictionaries: Key-Value Pairs 🌟

A **dictionary** maps keys to values — like a real dictionary maps words to definitions. 

This is the **most important data structure for AI-102** because JSON (the format Azure APIs use) maps directly to Python dictionaries.

````python
# Creating a dictionary
language_result = {
    "language": "English",
    "iso_code": "en",
    "confidence": 0.95
}

print(language_result)
````

````python
# Accessing values by key
language_result = {
    "language": "English",
    "iso_code": "en",
    "confidence": 0.95
}

print(language_result["language"])     # Get value by key
print(language_result["confidence"])   # Get another value
````

````python
# Safe access with .get() — won't crash if key doesn't exist
language_result = {
    "language": "English",
    "confidence": 0.95
}

# This is safer — returns None (or a default) if key is missing
print(language_result.get("language"))      # "English"
print(language_result.get("dialect"))       # None (key doesn't exist)
print(language_result.get("dialect", "N/A"))  # "N/A" (custom default)
````

````python
# Adding and updating values
config = {
    "endpoint": "https://myservice.cognitiveservices.azure.com",
    "api_key": "abc123"
}

# Add a new key
config["region"] = "eastus"

# Update an existing key
config["api_key"] = "new-key-456"

print(config)
````

## 📌 2.4 — Looping Through Dictionaries

When processing Azure AI responses, you often need to loop through all the key-value pairs.

````python
# Loop through keys and values
analysis = {
    "sentiment": "positive",
    "confidence": 0.89,
    "language": "en"
}

for key, value in analysis.items():
    print(f"  {key}: {value}")
````

## 📌 2.5 — Nested Structures (The Real AI-102 Pattern)

Azure AI responses are usually **dictionaries containing lists containing more dictionaries**. This nesting is the pattern you'll work with most.

Here's what a typical Azure AI response looks like:

```
response (dict)
├── "documents" (list)
│   ├── [0] (dict)
│   │   ├── "id": "1"
│   │   ├── "sentiment": "positive"
│   │   └── "confidenceScores" (dict)
│   │       ├── "positive": 0.89
│   │       ├── "neutral": 0.08
│   │       └── "negative": 0.03
│   └── [1] (dict)
│       ├── "id": "2"
│       └── ...
└── "modelVersion": "2024-01-01"
```

````python
# Simulating an Azure AI Language response
response = {
    "documents": [
        {
            "id": "1",
            "sentiment": "positive",
            "confidenceScores": {
                "positive": 0.89,
                "neutral": 0.08,
                "negative": 0.03
            }
        },
        {
            "id": "2",
            "sentiment": "negative",
            "confidenceScores": {
                "positive": 0.05,
                "neutral": 0.10,
                "negative": 0.85
            }
        }
    ],
    "modelVersion": "2024-01-01"
}

# Navigate through the nested structure
# Step 1: Get the documents list
documents = response["documents"]

# Step 2: Loop through each document
for doc in documents:
    doc_id = doc["id"]
    sentiment = doc["sentiment"]
    scores = doc["confidenceScores"]
    
    print(f"Document {doc_id}:")
    print(f"  Sentiment: {sentiment}")
    print(f"  Positive:  {scores['positive']:.0%}")
    print(f"  Neutral:   {scores['neutral']:.0%}")
    print(f"  Negative:  {scores['negative']:.0%}")
    print()
````

## 📌 2.6 — List Comprehensions (Quick & Clean)

A compact way to create lists. You'll sometimes see this in AI-102 lab code.

````python
# Traditional way
scores = [0.95, 0.67, 0.88, 0.45, 0.91]
high_scores = []
for s in scores:
    if s > 0.80:
        high_scores.append(s)
print(f"High confidence results: {high_scores}")

# Same thing with list comprehension (one line!)
high_scores = [s for s in scores if s > 0.80]
print(f"High confidence results: {high_scores}")
````
