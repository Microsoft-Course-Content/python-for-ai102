## 📌 4.1 — Reading Text Files

Some AI-102 labs read text files to analyze their content (sentiment, key phrases, etc.).

````python
# Writing a sample text file first (so we can read it)
with open("sample_review.txt", "w") as f:
    f.write("The hotel was amazing! Great service and beautiful rooms.\n")
    f.write("I would definitely recommend it to everyone.\n")
    f.write("The breakfast buffet was incredible.")

print("✅ File created!")
````

````python
# Reading the entire file at once
with open("sample_review.txt", "r") as f:
    content = f.read()

print("File Contents:")
print(content)
print(f"\nTotal characters: {len(content)}")
````

````python
# Reading line by line — useful for processing multiple reviews
with open("sample_review.txt", "r") as f:
    lines = f.readlines()

print(f"Number of lines: {len(lines)}")
for i, line in enumerate(lines, 1):
    print(f"  Line {i}: {line.strip()}")
````

## 📌 4.2 — The .env File Pattern 🔑

In AI-102 labs, your Azure endpoint and API key are stored in a `.env` file. This keeps sensitive information separate from your code.

The `.env` file looks like this:
```
AI_SERVICE_ENDPOINT=https://myservice.cognitiveservices.azure.com
AI_SERVICE_KEY=1a2b3c4d5e6f7g8h9i0j
```

The `python-dotenv` package loads these into environment variables.

````python
# Simulating the .env file pattern used in AI-102 labs

# Step 1: Create a .env file (in labs, this is already provided)
with open(".env", "w") as f:
    f.write("AI_SERVICE_ENDPOINT=https://eastus.api.cognitive.microsoft.com\n")
    f.write("AI_SERVICE_KEY=abc123def456ghi789\n")
    f.write("AI_SERVICE_REGION=eastus\n")

print("✅ .env file created!")
````

````python
# Step 2: Load the .env file and read values
# In AI-102 labs, you use: from dotenv import load_dotenv
# Here we'll simulate it manually since dotenv may not be installed

import os

# Manually loading .env (simulating what load_dotenv() does)
with open(".env", "r") as f:
    for line in f:
        line = line.strip()
        if line and "=" in line and not line.startswith("#"):
            key, value = line.split("=", 1)
            os.environ[key] = value

# Now read the values — THIS is the pattern you see in every lab
endpoint = os.getenv("AI_SERVICE_ENDPOINT")
key = os.getenv("AI_SERVICE_KEY")
region = os.getenv("AI_SERVICE_REGION")

print(f"Endpoint: {endpoint}")
print(f"Key: {key[:8]}...")  # Show only first 8 chars
print(f"Region: {region}")
````

### 💡 What `load_dotenv()` Does

In real AI-102 lab code, you'll see this cleaner version:

```python
from dotenv import load_dotenv
import os

# Load the .env file into environment variables
load_dotenv()

# Read values
endpoint = os.getenv("AI_SERVICE_ENDPOINT")
key = os.getenv("AI_SERVICE_KEY")
```

That's it! The `load_dotenv()` function reads your `.env` file and makes all the values available through `os.getenv()`.

## 📌 4.3 — Working with JSON Files

Some labs save or read JSON configuration files. JSON and Python dictionaries look almost identical.

````python
import json

# Writing JSON to a file
config = {
    "service": "TextAnalytics",
    "version": "v3.1",
    "features": ["sentiment", "keyPhrases", "entities"],
    "settings": {
        "language": "en",
        "showStats": True
    }
}

with open("config.json", "w") as f:
    json.dump(config, f, indent=2)

print("✅ config.json saved!")
````

````python
import json

# Reading JSON from a file
with open("config.json", "r") as f:
    loaded_config = json.load(f)

print(f"Service: {loaded_config['service']}")
print(f"Features: {loaded_config['features']}")
print(f"Language: {loaded_config['settings']['language']}")
````

## 📌 4.4 — Working with Image Files (Binary Files)

In AI-102 Vision labs, you read image files and send them to Azure for analysis. Images are **binary** files, so you open them with `"rb"` (read binary).

````python
# Creating a dummy image file for demonstration
# (In real labs, you'd have actual .jpg or .png files)
dummy_image_data = b"\x89PNG\r\n\x1a\n" + b"\x00" * 100  # Fake PNG header

with open("sample_image.png", "wb") as f:
    f.write(dummy_image_data)

print("✅ Dummy image created!")

# Reading an image file in binary mode — the AI-102 pattern
with open("sample_image.png", "rb") as f:
    image_data = f.read()

print(f"Image size: {len(image_data)} bytes")
print(f"First bytes: {image_data[:8]}")
````

## 📌 4.5 — Processing Multiple Files in a Folder

Some labs process all files in a folder — for example, analyzing sentiment for all review files in a directory.

````python
import os

# Create sample review files
os.makedirs("reviews", exist_ok=True)
reviews = {
    "review1.txt": "Excellent product! Works perfectly.",
    "review2.txt": "Terrible experience. Very disappointed.",
    "review3.txt": "It's okay. Nothing special."
}

for filename, content in reviews.items():
    with open(os.path.join("reviews", filename), "w") as f:
        f.write(content)

print("✅ Sample review files created!")
````

````python
import os

# Process all .txt files in the reviews folder
# This pattern appears in AI-102 labs that batch-process files

folder = "reviews"

for filename in os.listdir(folder):
    if filename.endswith(".txt"):
        filepath = os.path.join(folder, filename)
        
        with open(filepath, "r") as f:
            text = f.read()
        
        print(f"📄 {filename}")
        print(f"   Content: {text}")
        print(f"   Length: {len(text)} characters")
        print()
````

## 📌 4.6 — Putting It All Together: AI-102 Style

Let's combine everything into a script that follows the actual AI-102 lab structure.

````python
import os
import json

# ---- Step 1: Load configuration (simulating load_dotenv) ----
# In real labs: load_dotenv()
os.environ["AI_SERVICE_ENDPOINT"] = "https://eastus.api.cognitive.microsoft.com"
os.environ["AI_SERVICE_KEY"] = "your-api-key-here"

endpoint = os.getenv("AI_SERVICE_ENDPOINT")
key = os.getenv("AI_SERVICE_KEY")

print(f"🔗 Endpoint: {endpoint}")
print(f"🔑 Key loaded: {'Yes' if key else 'No'}")
print()

# ---- Step 2: Read files to process ----
folder = "reviews"
results = []

for filename in os.listdir(folder):
    if filename.endswith(".txt"):
        filepath = os.path.join(folder, filename)
        with open(filepath, "r") as f:
            text = f.read()
        
        # Step 3: Simulate API call (in real labs, this calls Azure)
        # Simplified simulation
        if "excellent" in text.lower() or "perfect" in text.lower():
            sentiment = "positive"
        elif "terrible" in text.lower() or "disappoint" in text.lower():
            sentiment = "negative"
        else:
            sentiment = "neutral"
        
        results.append({
            "file": filename,
            "text": text,
            "sentiment": sentiment
        })

# ---- Step 4: Display results ----
print("📊 Analysis Results:")
print("-" * 40)
for r in results:
    emoji = {"positive": "😊", "negative": "😞", "neutral": "😐"}
    print(f"  {r['file']}: {r['sentiment']} {emoji.get(r['sentiment'], '')}")

# ---- Step 5: Save results to JSON ----
with open("results.json", "w") as f:
    json.dump(results, f, indent=2)

print(f"\n✅ Results saved to results.json")
````
