## 📌 3.1 — Defining Functions

A **function** is a named block of code that you can call whenever you need it. Instead of writing the same code over and over, you write it once inside a function.

````python
# A simple function
def greet():
    print("Welcome to Azure AI Services! 🎉")

# Call the function
greet()
greet()  # You can call it as many times as you want
````

````python
# Function with a parameter (input)
def greet_user(name):
    print(f"Hello, {name}! Ready to explore AI?")

greet_user("Priya")
greet_user("Ahmed")
````

````python
# Function with a return value (output)
def calculate_percentage(score):
    return score * 100

confidence = 0.9321
result = calculate_percentage(confidence)
print(f"Confidence: {result:.1f}%")
````

## 📌 3.2 — Functions with Multiple Parameters

AI-102 functions often take several inputs — endpoint, key, text, options, etc.

````python
# Function with multiple parameters
def analyze_sentiment(text, language="en"):
    """Simulate sentiment analysis."""
    # This is a simulation — real AI-102 code calls Azure here
    if "great" in text.lower() or "love" in text.lower():
        sentiment = "positive"
        score = 0.92
    elif "bad" in text.lower() or "hate" in text.lower():
        sentiment = "negative"
        score = 0.88
    else:
        sentiment = "neutral"
        score = 0.75
    
    return {
        "text": text,
        "language": language,
        "sentiment": sentiment,
        "confidence": score
    }

# Call the function
result = analyze_sentiment("I love Azure AI Services!")
print(f"Sentiment: {result['sentiment']} ({result['confidence']:.0%})")

# Call with optional language parameter
result2 = analyze_sentiment("C'est magnifique!", language="fr")
print(f"Sentiment: {result2['sentiment']} ({result2['language']})")
````

## 📌 3.3 — The `if/elif/else` Pattern

Decision-making in code. You'll use this to handle different API responses, check thresholds, etc.

````python
# Basic if/else
confidence = 0.85

if confidence >= 0.90:
    print("🟢 High confidence")
elif confidence >= 0.70:
    print("🟡 Medium confidence")
else:
    print("🔴 Low confidence")
````

````python
# Real AI-102 pattern: Handling different sentiments
sentiment = "positive"

if sentiment == "positive":
    emoji = "😊"
elif sentiment == "negative":
    emoji = "😞"
elif sentiment == "neutral":
    emoji = "😐"
else:
    emoji = "❓"

print(f"Detected sentiment: {sentiment} {emoji}")
````

````python
# Combining conditions with 'and' / 'or'
confidence = 0.85
language = "en"

if confidence > 0.80 and language == "en":
    print("✅ High-confidence English result")

if language == "en" or language == "fr":
    print("✅ Supported language")
````

## 📌 3.7 — The `json` Module (Parsing API Responses)

**JSON** (JavaScript Object Notation) is the format Azure AI Services use to send and receive data. Python's `json` module converts between JSON strings and Python dictionaries.

````python
import json

# JSON string → Python dictionary (json.loads)
json_string = '{"language": "English", "confidence": 0.95, "isReliable": true}'

data = json.loads(json_string)  # Parse JSON string into dict
print(type(data))               # <class 'dict'>
print(data["language"])         # English
print(data["confidence"])       # 0.95
````

````python
import json

# Python dictionary → JSON string (json.dumps)
result = {
    "sentiment": "positive",
    "scores": {"positive": 0.89, "neutral": 0.08, "negative": 0.03}
}

# Convert to formatted JSON string
json_output = json.dumps(result, indent=2)
print(json_output)
````

## 📌 3.8 — Error Handling with try/except 🛡️

Things can go wrong — network errors, invalid API keys, missing files. The `try/except` pattern lets your code handle errors gracefully instead of crashing.

Every AI-102 lab uses this pattern.

````python
# Without error handling — the program crashes!
# number = int("hello")  # ← This would crash with ValueError

# With error handling — the program continues
try:
    number = int("hello")
    print(f"Number: {number}")
except ValueError:
    print("⚠️ That's not a valid number!")

print("Program continues running... ✅")
````

````python
# The AI-102 error handling pattern
def call_ai_service(text):
    """Simulate calling an Azure AI service with error handling."""
    try:
        # Simulate a potential error
        if not text:
            raise ValueError("Text cannot be empty")
        
        # Simulate successful API call
        result = {"sentiment": "positive", "confidence": 0.92}
        return result
        
    except ValueError as ve:
        print(f"⚠️ Validation error: {ve}")
        return None
    except Exception as ex:
        # This catches ANY other error
        print(f"❌ Unexpected error: {ex}")
        return None

# Test it
result1 = call_ai_service("Hello world!")
print(f"Result 1: {result1}")

result2 = call_ai_service("")  # Empty text
print(f"Result 2: {result2}")
````

````python
# try/except/finally — 'finally' always runs, even after an error
print("Opening connection to Azure AI service...")

try:
    # Simulating an API call
    print("Sending request...")
    result = {"status": "success"}
    print(f"Response: {result}")
except Exception as ex:
    print(f"Error: {ex}")
finally:
    # This ALWAYS runs — good for cleanup
    print("Closing connection... ✅")
````
