## 📌 1.3 — Data Types: What Kind of Value?

Python has a few core data types you'll encounter constantly in AI-102 labs:

| Type | What it stores | Example | AI-102 Usage |
|------|---------------|---------|--------------|
| `str` | Text | `"hello"` | API endpoints, keys, text to analyze |
| `int` | Whole numbers | `42` | Confidence scores, counts |
| `float` | Decimal numbers | `0.95` | Confidence percentages |
| `bool` | True/False | `True` | Feature flags, checks |

You can check a variable's type using `type()`.

````python
# Let's see the types in action
endpoint = "https://myservice.cognitiveservices.azure.com"
confidence = 0.97
word_count = 150
is_positive = True

print(type(endpoint))    # <class 'str'>
print(type(confidence))  # <class 'float'>
print(type(word_count))  # <class 'int'>
print(type(is_positive)) # <class 'bool'>
````

## 📌 1.4 — Strings: The Most Common Type in AI-102

Strings are text enclosed in quotes. In AI-102, almost everything starts as a string — your API key, the text you send for analysis, the JSON response you get back.

Let's learn the essential string operations.

````python
# Strings can use single or double quotes — both are fine
greeting = "Hello"
language = 'Python'

print(greeting)
print(language)
````

````python
# String concatenation (joining strings together)
first = "Azure"
second = "AI"
full = first + " " + second
print(full)
````

````python
# String length — useful when checking text limits
text = "Analyze this sentence for sentiment"
print(len(text))  # Number of characters
````

````python
# String methods you'll see in AI-102 labs
api_key = "  AbC123xYz  "

print(api_key.strip())    # Remove extra spaces
print(api_key.lower())    # All lowercase  
print(api_key.upper())    # All uppercase

# Checking content
url = "https://eastus.api.cognitive.microsoft.com"
print(url.startswith("https"))  # True
print(url.endswith(".com"))     # True
````

## 📌 1.5 — f-Strings: The AI-102 Lab Favorite 🌟

**f-strings** (formatted strings) let you embed variables directly inside a string. You'll see these *everywhere* in AI-102 lab code.

The syntax is simple: put an `f` before the quote, and wrap variables in `{curly braces}`.

````python
# Basic f-string
name = "Computer Vision"
version = "4.0"
print(f"Using {name} API version {version}")
````

````python
# f-strings are used heavily in AI-102 for building URLs and printing results
endpoint = "https://myservice.cognitiveservices.azure.com"
api_version = "2024-02-01"

# Building an API URL — you'll see this pattern in every AI-102 lab!
url = f"{endpoint}/language/:analyze-text?api-version={api_version}"
print(url)
````

````python
# f-strings with expressions
confidence = 0.9432
print(f"Confidence: {confidence:.2f}")     # Format to 2 decimal places
print(f"Confidence: {confidence:.0%}")      # Format as percentage
print(f"Confidence: {confidence * 100:.1f}%")  # Manual percentage
````

## 📌 1.7 — Basic Math Operations

You'll need simple math for working with confidence scores, percentages, and counting results.

````python
# Basic arithmetic
a = 10
b = 3

print(f"Addition:       {a + b}")
print(f"Subtraction:    {a - b}")
print(f"Multiplication: {a * b}")
print(f"Division:       {a / b:.2f}")    # Regular division (float)
print(f"Floor Division: {a // b}")        # Rounds down to integer
print(f"Remainder:      {a % b}")         # Modulo
print(f"Power:          {a ** b}")         # 10 to the power of 3
````

````python
# Real-world AI-102 example: Working with confidence scores
confidence = 0.8765
threshold = 0.75

# Convert to percentage
percentage = confidence * 100
print(f"Confidence: {percentage:.1f}%")

# Check if above threshold
print(f"Above threshold: {confidence > threshold}")
````
