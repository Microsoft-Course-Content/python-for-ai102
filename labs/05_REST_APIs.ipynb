## 📌 5.1 — HTTP Methods: The Basics

REST APIs use different HTTP methods for different actions:

| Method | What it does | AI-102 Usage |
|--------|-------------|--------------|
| `GET` | Retrieve data | Get service status, list models |
| `POST` | Send data for processing | Analyze text, detect objects |
| `PUT` | Create or update | Deploy models, create resources |
| `DELETE` | Remove something | Delete resources |

In AI-102, you'll mostly use **POST** (sending data for AI analysis) and **GET** (retrieving results).

## 📌 5.3 — Anatomy of an API Request

Every API call in AI-102 has these parts:

```
┌─────────────────────────────────────────────────┐
│                  API REQUEST                     │
│                                                  │
│  URL:     https://endpoint/service/analyze       │
│           ▲                                      │
│           └── Where to send the request          │
│                                                  │
│  HEADERS: {                                      │
│    "Ocp-Apim-Subscription-Key": "your-key",     │
│    "Content-Type": "application/json"            │
│  }                                               │
│  ▲                                               │
│  └── Authentication + format info                │
│                                                  │
│  BODY:    {                                      │
│    "documents": [                                │
│      {"id": "1", "text": "Hello world"}          │
│    ]                                             │
│  }                                               │
│  ▲                                               │
│  └── The data you want analyzed                  │
└─────────────────────────────────────────────────┘
```

````python
import json

# Let's build an API request step by step — AI-102 style

# Step 1: Define the endpoint and key
endpoint = "https://myservice.cognitiveservices.azure.com"
key = "your-subscription-key"

# Step 2: Build the URL for the specific service
url = f"{endpoint}/text/analytics/v3.1/sentiment"

# Step 3: Set up headers (authentication + content type)
headers = {
    "Ocp-Apim-Subscription-Key": key,     # Azure API Key
    "Content-Type": "application/json"       # We're sending JSON
}

# Step 4: Build the request body
body = {
    "documents": [
        {"id": "1", "language": "en", "text": "I love this product!"},
        {"id": "2", "language": "en", "text": "This is terrible."},
        {"id": "3", "language": "en", "text": "It's fine, nothing special."}
    ]
}

# Step 5: (Simulated) Make the request
# In real code: response = requests.post(url, headers=headers, json=body)

print("📤 Request prepared:")
print(f"  URL: {url}")
print(f"  Headers: {json.dumps(headers, indent=4)}")
print(f"  Body: {json.dumps(body, indent=4)}")
````

## 📌 5.4 — Handling API Responses

When you make an API call, you get back a **response object** with a status code and data.

````python
import json

# Simulating what an Azure API response looks like
# In real code: response = requests.post(url, headers=headers, json=body)

# The response has these key attributes:
class SimulatedResponse:
    status_code = 200
    
    def json(self):
        return {
            "documents": [
                {
                    "id": "1",
                    "sentiment": "positive",
                    "confidenceScores": {"positive": 0.98, "neutral": 0.01, "negative": 0.01}
                },
                {
                    "id": "2",
                    "sentiment": "negative",
                    "confidenceScores": {"positive": 0.02, "neutral": 0.05, "negative": 0.93}
                },
                {
                    "id": "3",
                    "sentiment": "neutral",
                    "confidenceScores": {"positive": 0.15, "neutral": 0.80, "negative": 0.05}
                }
            ],
            "modelVersion": "2024-01-01"
        }

# Process the response — THIS is what you'll do in labs
response = SimulatedResponse()

# Check if the request was successful
if response.status_code == 200:
    results = response.json()
    
    # Process each document
    for doc in results["documents"]:
        doc_id = doc["id"]
        sentiment = doc["sentiment"]
        scores = doc["confidenceScores"]
        
        # Find the highest score
        max_score = max(scores.values())
        
        print(f"Document {doc_id}: {sentiment} (confidence: {max_score:.0%})")
else:
    print(f"❌ Error: Status code {response.status_code}")
````

## 📌 5.5 — HTTP Status Codes You'll Encounter

| Code | Meaning | What to do |
|------|---------|------------|

| `400` | ❌ Bad Request | Check your request body |
| `401` | ❌ Unauthorized | Check your API key |
| `403` | ❌ Forbidden | Check your permissions |
| `404` | ❌ Not Found | Check the URL |
| `429` | ⚠️ Too Many Requests | Slow down, add a delay |
| `500` | ❌ Server Error | Try again later |
